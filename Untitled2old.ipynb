```python
import glob, os, numpy as np
import trimesh as tm

DATA_DIR = "/content/PartField/data/my_objs"
bad = []
for p in sorted(glob.glob(os.path.join(DATA_DIR, "*"))):
    try:
        m = tm.load(p, force='mesh')
        V = np.asarray(getattr(m, "vertices", np.empty((0,3))))
        F = np.asarray(getattr(m, "faces", np.empty((0,3))))
        # mark as bad if empty, or no faces
        if V.size == 0 or F.size == 0:
            print("BAD (empty):", p, V.shape, F.shape)
            bad.append(p)
        else:
            print("OK :", p, V.shape, F.shape)
    except Exception as e:
        print("ERR:", p, "->", e)
        bad.append(p)

print("\nBad/errored files:", len(bad))
```

```python
import os, shutil
QUAR = "/content/PartField/data/quarantine"
os.makedirs(QUAR, exist_ok=True)
for p in bad:
    shutil.move(p, os.path.join(QUAR, os.path.basename(p)))
print("Quarantined:", len(bad), "files to", QUAR)
```

```python
import os, glob

# expected paths
ROOT = "/content/PartField/exp_results/partfield_features/my_objs_demo"
SRC = "/content/PartField/data/my_objs"

print("🔎 Checking features in:", ROOT)
npy_files = sorted(glob.glob(os.path.join(ROOT, "part_feat_*.npy")))
print("  Found", len(npy_files), "feature files")
for f in npy_files[:5]:
    print("   ", os.path.basename(f))
if not npy_files:
    print("  ⚠️ No features found!")

print("\n🔎 Checking meshes in:", SRC)
mesh_files = sorted(glob.glob(os.path.join(SRC, "*.obj")))
print("  Found", len(mesh_files), "meshes")
for f in mesh_files[:5]:
    print("   ", os.path.basename(f))
if not mesh_files:
    print("  ⚠️ No meshes found!")

# quick consistency check
missing = []
for f in npy_files:
    base = os.path.basename(f).replace("part_feat_", "").split("_0")[0]
    match = [m for m in mesh_files if base in m]
    if not match:
        missing.append(base)

if missing:
    print("\n⚠️ Features with no matching mesh:", len(missing))
    print(missing[:5])
else:
    print("\n✅ All features have matching meshes (at least by name fragment)")
```

```python
import os, glob, trimesh as tm

SRC = "/content/PartField/data/my_objs"
FEAT = "/content/PartField/exp_results/partfield_features/my_objs_demo"

os.makedirs(FEAT, exist_ok=True)

made, skipped = 0, 0
for obj in sorted(glob.glob(os.path.join(SRC, "*.obj"))):
    stem = os.path.splitext(os.path.basename(obj))[0]
    ply_out = os.path.join(FEAT, f"input_{stem}_0.ply")
    if os.path.exists(ply_out):
        skipped += 1
        print("exists:", os.path.basename(ply_out))
        continue
    try:
        mesh = tm.load(obj, force='mesh', process=False)
        # guard against empty meshes
        if mesh.is_empty or (getattr(mesh, "vertices", None) is None) or len(mesh.vertices) == 0:
            print("skip empty:", obj)
            continue
        mesh.export(ply_out)
        made += 1
        print("made  :", os.path.basename(ply_out))
    except Exception as e:
        print("error :", obj, "->", e)

print(f"\nDone. made={made}, skipped={skipped}")
```
